# GA Energy Inspection

This notebook is the browser-native companion to the mdBook GA chapter.

It is meant for interactive inspection and plotting, not for launching native `cargo run` commands.

````python
import json
import plotly.graph_objects as go

ga = {
    "generation": [0, 1, 2, 3, 4],
    "best_energy": [-14.82, -15.21, -15.67, -15.74, -15.91],
    "mean_energy": [-14.10, -14.44, -14.88, -15.02, -15.19],
    "duplicates": [0, 1, 2, 1, 1],
    "repopulated": [0, 0, 1, 0, 1],
}

print(json.dumps(ga, indent=2))
````

````python
fig = go.Figure()
fig.add_scatter(x=ga['generation'], y=ga['best_energy'], mode='lines+markers', name='Best energy')
fig.add_scatter(x=ga['generation'], y=ga['mean_energy'], mode='lines+markers', name='Mean energy')
fig.add_bar(x=ga['generation'], y=ga['duplicates'], name='Duplicates', opacity=0.25, yaxis='y2')
fig.add_scatter(x=ga['generation'], y=ga['repopulated'], mode='markers', name='Repopulated', yaxis='y2')

fig.update_layout(
    title='GA generation metrics',
    xaxis_title='Generation',
    yaxis_title='Energy (eV)',
    yaxis2=dict(title='Operational count', overlaying='y', side='right', rangemode='tozero')
)

fig.show()
````
